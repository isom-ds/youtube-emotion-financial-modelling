## Returns using API

```python
import yfinance as yf
```

```python
data = yf.download(
    [
        '^GSPC', 'GC=F', 'BTC-USD', 
        'EUR=X', 'CNY=X', '^TNX', 'CL=F', 
        '^SP500-2510', '^SP500-101020', '^SP500-201060', '^SP500-452030',
        'AAPL',
    ],
    period='2y', interval='1d'
)
```

```python
# Export the downloaded 'data' variable to JSON in the repo 'data/' folder
import os
import json
import pandas as pd

df = data.copy()
# If DataFrame has a DatetimeIndex (common with yfinance), make it a column
try:
    df = df.reset_index()
except Exception:
    pass
# Flatten MultiIndex columns (yfinance returns MultiIndex columns)
if isinstance(df.columns, pd.MultiIndex):
    df.columns = ['_'.join(map(str, col)).strip() for col in df.columns.values]
# Convert to list-of-records for JSON serialization
out = df.to_dict(orient='records')
os.makedirs('data', exist_ok=True)
out_path = os.path.join('data', 'market_data.json')
with open(out_path, 'w') as f:
    json.dump(out, f, default=str, indent=2)
print(f'Saved data to {out_path}')
```
